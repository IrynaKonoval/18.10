- Для категорії Food and beverages: члени клубу — основні покупці, тож варто утримувати їх персональними пропозиціями на продукти.
- Для Electronic accessories: багато покупців без картки. Можна пропонувати
  оформити картку саме при покупці електроніки (наприклад, знижка на першу покупку), щоб залучати цих клієнтів у програму.

2. Коли найбільше купують: день тижня

**Гіпотеза: у вихідні виручка вища, ніж у будні.**

```python
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Weekday'] = df['Date'].dt.day_name()

df.groupby('Weekday')['Total'].agg(['count', 'sum', 'mean']).sort_values('sum', ascending=False)
```

```python
df['Date'].min(), df['Date'].max()
```

**Висновки:** гіпотеза підтвердилась **частково**.
- Субота — найсильніший день: 164 покупки, виручка 56 121, найвищий середній чек (342).
- Неділя — лише 4-та за виручкою, але з високим середнім чеком (334):
  покупців менше, проте вони купують більше.
- Несподівано сильний вівторок — 2-ге місце за виручкою (51 482).
- Понеділок — найслабший день за всіма показниками (виручка на 48% нижча, ніж у суботу).

**Ревоиендації:**
- **Субота:** виводити більше персоналу й кас, заздалегідь поповнювати запаси.
- **Понеділок:** запускати акції, щоб залучити покупців
  (наприклад, «знижка понеділка» або подвійні бонуси для членів клубу).
- **Вівторок:** з'ясувати причину піку (можливо, у цей день завозять товар
  або діє якась акція) і повторити цей досвід в інші будні.
- **Неділя:** працювати над тим, щоб покупців було більше, адже чек уже високий.

**Обмеження:** дані охоплюють лише 3 місяці (січень–березень 2019),
тобто приблизно по 12–13 тижнів на кожен день, тому висновки варто
перевірити на даних за довший період.

**Гіпотеза: чоловіки й жінки витрачають більше на різні категорії товарів.**

```python
gender_category.sum()
```

**Висновки:** гіпотеза підтвердилась.
- **Жінки** витрачають більше на Food and beverages, Fashion accessories
  і Home and lifestyle. Загалом жінки витратили більше
  167 900 проти 155 100, +8%.
- **Чоловіки** несподівано лідирують у категорії Health and beauty
  (30 633 проти 18 561).
- **Electronic accessories** купують порівну.

**Рекомендації:**

**Обмеження:** вибірка невелика (1000 покупок за 3 місяці),
